```python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../DataSet/diamonds.csv")
```

```python
data.head()
```

```python
data = data.iloc[:,1:]
```

```python
data.head()
```

```python
plt.plot(data['carat'], data["price"])
```

```python
plt.scatter(data['carat'], data['price'])
```

```python
plt.bar(data["cut"].unique(), data.groupby("cut")["price"].mean())
```

```python
plt.hist(data["carat"], bins=110, edgecolor="black")
plt.show()
```

```python
plt.boxplot(data["carat"], vert=False)
plt.show()
```

```python
cut_count = data["cut"].value_counts()
```

```python
plt.pie(cut_count, labels=cut_count.index, autopct="%1.1f%%", startangle=90)
plt.show()
```

```python
data.head()
```

```python
corr = data[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']].corr()
corr
plt.imshow(corr, cmap="coolwarm", interpolation="none")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
```
