<h2>Grammer of Graphics for plotting in Python</h2>
<ul>
  <li><b>Data:</b> The data to be visualized</li>
  <li><b>Aesthetics:</b> The visual properties of the data points, such as position, size, shap, color and transparency</li>
  <li><b>Geometry:</b> The geometric shapes used to represent the data points such as bars, line, and points</li>
  <li><b>Facets:</b> The arrangement of the data into subset such os rows, columns or panels</li>
  <li><b>Statistics:</b> The statistical transformations applied to the data such as aggregations, smoothing and modeling</li>
  <li><b>Scales:</b> The Mapping of data values to visual properties such as axis limits, tick marks and color scales</li>
  <li><b>Themes:</b> The overall visual appearance of the plot such as colors, fonts and gridlines</li>
  <li><b>Guides:</b> The visual elements used to assist in reading the plot such as legends, axes and labels</li>
  <li><b>Interactivity:</b> The interactive features of the plot such as tooltips, zooming and panning</li>
  <li><b>Annotations:</b> The textual and graphical annotations used to highlight specific aspects of the plot</li>
  <li><b>Extensions:</b> The additional features and plugins that enhance the plot such as animations, sliders and widgents</li>
</ul>

## Data Visualization with Matplotlib and Seaborn

### Loading the Libraries and data

```python
# load the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data = pd.read_csv("../DataSet/diamonds.csv", delimiter=",")
```

```python
data.head()
```

```python
data.drop(columns=["Unnamed: 0"], inplace=True)
```

## Explain each part of diamonds dataset

<p>A data frame which 53,940 rows and 10 variables</p>

<ui>
    <li><b>price: </b>price in US dollars (\$326--\$18,823) </li>
    <li><b>carat: </b>weight of the diamond (0.2--5.01) </li>
    <li><b>cut: </b> quality of the cut (Fair, Good, Very Good, Premium, Ideal) </li>
    <li><b>color: </b>diamond colour, from J (worst) to D (best) </li>
    <li><b>clarity: </b>a measurement of how clear the diamond is (I1 (worst), SI2, SI1, VS2, VS1, VVS2, VVS1, IF (best)) </li>
    <li><b>x: </b>length in mm (0--10.74) </li>
    <li><b>y: </b>width in mm (0--58.9)</li>
    <li><b>z: </b>depth in mm (0--31.8) </li>
    <li><b>depth: </b>total depth percentage = z / mean(x, y) = 2 * z / (x + y) (43--79) </li>
    <li><b>table: </b>width of top of diamond relative to widest point (43--95) </li>
</ui>

```python
plt.plot(data["carat"], data["price"])
plt.title("Carat vs Price Line Plot")
plt.xlabel("Carat")
plt.ylabel("Price")
plt.show()
```

```python
plt.scatter(data["carat"], data["price"])
plt.title("Carat vs Price Scatter Plot")
plt.xlabel("Carat")
plt.ylabel("Price")
plt.show()
```

```python
plt.bar(data["cut"].unique(), data.groupby("cut")["price"].mean())
plt.title("Average Price By cut")
plt.xlabel("Cut Types")
plt.ylabel("Average Price")
plt.show()
```
